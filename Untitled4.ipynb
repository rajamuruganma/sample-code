```python
stack = []
top = -1

def push(data):
    global top
    stack.append(data)
    top += 1
    print(data, "inserted")

def pop():
    global top
    if top == -1:
        print("Stack is Empty")
    else:
        print("Deleted element:", stack[top])
        stack.pop()
        top -= 1

def peek():
    if top == -1:
        print("Stack is Empty")
    else:
        print("Top element:", stack[top])

push(10)
push(20)
push(30)

print("Stack:", stack)
peek()

pop()

print("Stack after pop:", stack)
peek()
```

```python
queue = []
front = -1
rear = -1

def enqueue(item):
    global front, rear
    if front == -1:
        front = 0
    queue.append(item)
    rear += 1
def dequeue():
    global front, rear
    if front == -1 or front > rear:
        print("Queue Underflow")
        return
    item = queue[front]
    front += 1
    if front > rear:
        front = rear = -1
    return item

enqueue(10)
enqueue(20)
enqueue(30)


dequeue()
```
