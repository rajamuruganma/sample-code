```python
arr = [1, 2, 3, 5]

n = len(arr) + 1

total = n * (n + 1) // 2

s = 0
for i in arr:
    s += i

missing = total - s

print(missing)
```

```python
arr = [1, 2, 3, 4, 5]

start = 0
end = len(arr) - 1

while start < end:
    arr[start], arr[end] = arr[end], arr[start]

    start += 1
    end -= 1

print(arr)
```

```python
arr = [3, 2, 2, 3]
val = 3

i = 0
j = 0

while j < len(arr):

    if arr[j] != val:
        arr[i] = arr[j]
        i += 1

    j += 1

print(arr[:i])
print(i)
```
